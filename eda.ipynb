```python
# eda_pnc_claims.py
# End-to-end EDA for the FNOL-safe regression dataset

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import OrdinalEncoder, StandardScaler
from sklearn.feature_selection import mutual_info_regression
from sklearn.decomposition import PCA
from statsmodels.stats.outliers_influence import variance_inflation_factor
import statsmodels.api as sm

# ---------- Load
df = pd.read_csv("synthetic_pnc_claims_regression.csv", parse_dates=["Date_of_Loss","Report_Date"])

# ---------- Define schema (exact columns from our generator)
TARGET = "Total_Paid_Amount"

cat_cols = [
    "Coverage_Type","Insurance_Score_Band","Vehicle_Segment","Urbanicity",
    "Collision_Type","Weather_At_LOSS","At_Fault_Indicator"
]
bool_cols = ["Injury_Indicator","Multi_Vehicle_Flag","Attorney_Rep_At_FNOL"]
num_cols = [
    "Bodily_Injury_Limit_Per_Person","Property_Damage_Limit","Collision_Deductible",
    "Policy_Tenure_Months","Driver_Age",
    "Vehicle_Age_Years","Vehicle_ACV",
    "Repair_Cost_Index","Medical_Cost_Index","Weather_Cat_Risk_Index",
    "Report_Lag_Days","Prior_Claims_Count_36M","Moving_Violations_36M"
]
id_date_cols = ["Claim_ID","Policy_ID","Date_of_Loss","Report_Date"]

# Safety check: ensure columns exist
assert set([TARGET] + cat_cols + bool_cols + num_cols + id_date_cols).issubset(df.columns), "Column mismatch!"

# Cast types
for c in cat_cols:
    df[c] = df[c].astype("category")
for c in bool_cols:
    df[c] = df[c].astype(bool)
for c in num_cols:
    df[c] = pd.to_numeric(df[c], errors="coerce")

# ---------- 1) Quick overview
print("Shape:", df.shape)
print("\nDtypes:\n", df.dtypes)
print("\nMissing (%):\n", df.isna().mean().sort_values(ascending=False).head(20))

# ---------- 2) Target distribution (with and without log)
fig, ax = plt.subplots(1,2, figsize=(12,4))
sns.histplot(df[TARGET], ax=ax[0], bins=50)
ax[0].set_title("Total_Paid_Amount")
sns.histplot(np.log1p(df[TARGET]), ax=ax[1], bins=50)
ax[1].set_title("log1p(Total_Paid_Amount)")
plt.tight_layout(); plt.show()

print("\nTarget summary:\n", df[TARGET].describe(percentiles=[.5,.75,.9,.95,.99]))

# ---------- 3) Univariate – numerics
_ = df[num_cols].describe().T.assign(missing=df[num_cols].isna().mean())
print("\nNumeric summary (+ missing):\n", _)

# Optional: histograms for numerics (small multiples)
n = len(num_cols)
cols = 4
rows = int(np.ceil(n/cols))
fig, axes = plt.subplots(rows, cols, figsize=(4*cols, 3*rows))
axes = axes.ravel()
for i,c in enumerate(num_cols):
    sns.histplot(df[c], bins=40, ax=axes[i])
    axes[i].set_title(c)
for j in range(i+1, rows*cols):
    axes[j].axis("off")
plt.tight_layout(); plt.show()

# ---------- 4) Univariate – categoricals
for c in cat_cols + bool_cols:
    print(f"\n{c} value counts:")
    print(df[c].value_counts(dropna=False, normalize=True).head(20))

# ---------- 5) Correlation (numerics)
corr_p = df[num_cols + [TARGET]].corr(method="pearson")
corr_s = df[num_cols + [TARGET]].corr(method="spearman")

plt.figure(figsize=(10,8))
sns.heatmap(corr_p, annot=False, cmap="coolwarm", center=0)
plt.title("Pearson Correlation (numerics + target)")
plt.show()

print("\nTop Pearson corr with target:")
print(corr_p[TARGET].sort_values(ascending=False))

print("\nTop Spearman corr with target:")
print(corr_s[TARGET].sort_values(ascending=False))

# ---------- 6) Target vs cat features (mean target by category)
def target_by_category(df, col, target=TARGET, top=15):
    agg = (df.groupby(col)[target]
           .agg(["count","mean","median","std"])
           .sort_values("mean", ascending=False))
    print(f"\n{col} → target stats:\n", agg.head(top))
    # bar (means)
    m = agg["mean"]
    m.plot(kind="bar", figsize=(8,3), title=f"{col} | mean {target}")
    plt.tight_layout(); plt.show()

for c in cat_cols + bool_cols:
    target_by_category(df, c)

# ---------- 7) Mutual Information (captures non-linearities & cat effects)
X_mi = df[cat_cols + bool_cols + num_cols].copy()
# Encode categoricals ordinally just for MI (no leakage - unsupervised transform)
enc = OrdinalEncoder(handle_unknown="use_encoded_value", unknown_value=-1)
X_cat = enc.fit_transform(df[cat_cols]) if cat_cols else np.empty((len(df),0))
X_bool = df[bool_cols].astype(int).to_numpy() if bool_cols else np.empty((len(df),0))
X_num  = df[num_cols].to_numpy()
X_all  = np.hstack([X_cat, X_bool, X_num])

mi = mutual_info_regression(X_all, df[TARGET], random_state=0)
mi_index = pd.Index(cat_cols + bool_cols + num_cols, name="feature")
mi_series = pd.Series(mi, index=mi_index).sort_values(ascending=False)
print("\nMutual Information (higher = more predictive):\n", mi_series.head(20))
mi_series.head(30).plot(kind="bar", figsize=(10,3), title="Mutual Information with Target")
plt.tight_layout(); plt.show()

# ---------- 8) Multicollinearity (VIF) — start with NUMERICS ONLY
# (Doing VIF after one-hot on many dummies is noisy; this flags numeric redundancy first.)
X_vif = df[num_cols].dropna().copy()
X_vif = (X_vif - X_vif.mean())/X_vif.std(ddof=0)  # standardize
X_vif = sm.add_constant(X_vif)  # intercept for statsmodels
vifs = pd.Series(
    [variance_inflation_factor(X_vif.values, i) for i in range(X_vif.shape[1])],
    index=X_vif.columns
).drop("const").sort_values(ascending=False)
print("\nVIF (numerics):\n", vifs)

# ---------- 9) PCA (numerics) — to inspect latent structure
X_num_std = StandardScaler().fit_transform(df[num_cols].fillna(df[num_cols].median()))
pca = PCA(n_components=min(10, len(num_cols)))
pca.fit(X_num_std)
expl = pd.Series(pca.explained_variance_ratio_, index=[f"PC{i+1}" for i in range(len(pca.explained_variance_ratio_))])
print("\nPCA explained variance (numerics):\n", expl.cumsum())

plt.figure(figsize=(6,3))
expl.plot(marker="o")
plt.title("Scree: variance explained per PC (numerics)")
plt.ylabel("Explained variance ratio"); plt.xlabel("Principal Component")
plt.tight_layout(); plt.show()

# Optional: project and visualize first two PCs against target (sample to avoid overplot)
proj = pca.transform(X_num_std)
sample_idx = np.random.choice(len(df), size=min(5000, len(df)), replace=False)
plt.figure(figsize=(5,4))
sc = plt.scatter(proj[sample_idx,0], proj[sample_idx,1], c=df[TARGET].iloc[sample_idx], s=6, alpha=0.6)
plt.colorbar(sc, label=TARGET)
plt.title("PC1 vs PC2 (colored by target)")
plt.tight_layout(); plt.show()

# ---------- 10) Bivariate quick looks (numeric vs target)
# Spearman scatter trend on a few top MI features
top_num_for_plots = [f for f in mi_series.index if f in num_cols][:6]
cols = 3
rows = int(np.ceil(len(top_num_for_plots)/cols)) if top_num_for_plots else 0
if rows > 0:
    fig, axes = plt.subplots(rows, cols, figsize=(5*cols, 3.5*rows))
    axes = np.atleast_1d(axes).ravel()
    for i,c in enumerate(top_num_for_plots):
        sns.regplot(x=df[c], y=df[TARGET], scatter_kws={"s":6, "alpha":0.3}, line_kws={"color":"k"}, ax=axes[i])
        axes[i].set_title(f"{c} vs {TARGET}")
    for j in range(i+1, rows*cols):
        axes[j].axis("off")
    plt.tight_layout(); plt.show()

# ---------- 11) Outlier & capping suggestion (report only)
q99 = df[TARGET].quantile(0.99)
q999 = df[TARGET].quantile(0.999)
print(f"\nTarget tail check: 99th={q99:,.0f}, 99.9th={q999:,.0f}. Consider log1p transform or capping.")

# ---------- 12) Feature-prep sketch (no modeling yet)
# Tip: encode cats with OneHot(handle_unknown='ignore'), scale numerics, leave booleans as 0/1.
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

ohe = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
scaler = StandardScaler()

preprocessor = ColumnTransformer(
    transformers=[
        ("num", scaler, num_cols),
        ("cat", ohe, cat_cols),
        ("bool", "passthrough", bool_cols),
    ],
    remainder="drop"
)

print("\nPreprocessor ready (ColumnTransformer with OHE + scaling). Next step: fit a model on train/valid split.")
```
